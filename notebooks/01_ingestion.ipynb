# 01 — Ingestion

Calls the real parsers in `ingest.py` on real files from the configured scan folder. No reimplemented parsing logic — every function here is imported from the actual engine.

```python
import sys, os
sys.path.insert(0, "..")
os.chdir("..")

from pathlib import Path
from dotenv import load_dotenv
load_dotenv()

from ingest import collect_files, PARSERS

scan_dir = Path(os.environ.get("LOCAL_SCAN_DIR", "~/Downloads")).expanduser()
files = collect_files(scan_dir)
print(f"{len(files)} supported files found under {scan_dir}")
```

Pick one real PDF and one real DOCX already in the corpus and run the actual parser functions on them.

```python
sample_pdf = next(f for f in files if f.name == "Unit 3 Finance and Marketing for early entrepreneurs.pdf")
sample_docx = next(f for f in files if f.suffix.lower() == ".docx" and "DPR" in f.name)

print("PDF sample:", sample_pdf.name)
print("DOCX sample:", sample_docx.name)
```

```python
pdf_sections = PARSERS[".pdf"](sample_pdf)
print(f"{sample_pdf.name}: {len(pdf_sections)} section(s) extracted")
text, location = pdf_sections[0]
print(f"--- {location} (first 400 chars) ---")
print(text[:400])
```

```python
docx_sections = PARSERS[".docx"](sample_docx)
print(f"{sample_docx.name}: {len(docx_sections)} section(s) extracted")
text, location = docx_sections[0]
print(f"--- {location} (first 400 chars) ---")
print(text[:400])
```
