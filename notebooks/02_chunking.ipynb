# 02 — Chunking

Calls the real `chunk_text()` from `ingest.py` (character-based, bug-fix #1) on real extracted text and shows the actual chunk boundaries and sizes.

```python
import sys, os
sys.path.insert(0, "..")
os.chdir("..")

from pathlib import Path
from dotenv import load_dotenv
load_dotenv()

from ingest import collect_files, PARSERS, chunk_text, CHUNK_CHARS, OVERLAP_RATIO

scan_dir = Path(os.environ.get("LOCAL_SCAN_DIR", "~/Downloads")).expanduser()
files = collect_files(scan_dir)
sample_docx = next(f for f in files if f.suffix.lower() == ".docx" and "DPR" in f.name)

sections = PARSERS[".docx"](sample_docx)
full_text, location = sections[0]
print(f"{sample_docx.name} / {location}: {len(full_text)} characters total")
print(f"CHUNK_CHARS={CHUNK_CHARS}, OVERLAP_RATIO={OVERLAP_RATIO}")
```

```python
chunks = chunk_text(full_text)
print(f"{len(chunks)} chunk(s) produced\n")

pos = 0
for i, c in enumerate(chunks):
    start = full_text.find(c[:40], pos)
    print(f"chunk {i+1}: {len(c)} chars, starts at char offset ~{start}")
    pos = max(pos, start)
```

```python
print("--- chunk 1, first 200 chars ---")
print(chunks[0][:200])
print()
print("--- chunk 2, first 200 chars (note the overlap with the tail of chunk 1) ---")
print(chunks[1][:200])
```
