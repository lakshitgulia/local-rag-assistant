# 03 — Embedding + Vector Store

Uses the real `get_embeddings()` from `core.py` and LangChain's `FAISS` vectorstore, loading the actual persisted index built by `ingest.py` against the real corpus.

```python
import sys, os
sys.path.insert(0, "..")
os.chdir("..")

from langchain_community.vectorstores import FAISS
from core import FAISS_DIR, get_embeddings, load_chunks

chunks = load_chunks()
print(f"{len(chunks)} chunks loaded from the real persisted index")

embeddings = get_embeddings()
vectorstore = FAISS.load_local(FAISS_DIR, embeddings, allow_dangerous_deserialization=True)
print("FAISS vectorstore loaded from", FAISS_DIR)
```

Embed a real question and run vector similarity search directly against the real index.

```python
question = "What is the formula for calculating CAGR?"
results = vectorstore.similarity_search_with_score(question, k=5)

for i, (doc, score) in enumerate(results):
    print(f"[{i+1}] score={score:.4f}  {doc.metadata['source_file']} / {doc.metadata['location']}")
    print("    ", doc.page_content[:120].replace(chr(10), " "))
```
